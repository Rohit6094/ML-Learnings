```python
import yfinance as yf
import pandas as pd
import talib as ta
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
tesla_df = yf.download(
    tickers='TSLA',
    period='5y',
    auto_adjust=True,
    interval='1d'
)
print(tesla_df.head())

tesla_df.to_csv('Tesla_Stock_5Years.csv')
```

```python
data = pd.read_csv('D:\\codes\\files\\ML-Learnings\\yfinance\\Tesla_Stock_5Years.csv')
df = pd.DataFrame(data)
df.head()
df.drop([0,1], inplace=True)
```

```python
df['Date'] = df['Price']
df.head()
df.drop(columns='Price', axis=1, inplace=True)
```

```python
df['Date'] = pd.to_datetime(df['Date'])

df.head()
```

```python
df['Close'] = pd.to_numeric(df['Close'], errors='coerce')
df['High'] = pd.to_numeric(df['High'], errors='coerce')
df['Low'] = pd.to_numeric(df['Low'], errors='coerce')
df['Open'] = pd.to_numeric(df['Open'], errors='coerce')
df['Volume'] = pd.to_numeric(df['Volume'], errors='coerce')

df.info()
```

```python
px.line(df, x='Date', y='Close', title='Tesla Stock Price for 5 Years')
```

```python
fig = make_subplots(
    rows=3, cols=1,shared_xaxes=False,
    vertical_spacing=0.05,
    subplot_titles=('OHLC', 'Volume', 'RSI'),
    row_heights=[0.5, 0.2, 0.3]
)

fig.add_trace(
    go.Candlestick(x=df['Date'], open=df['Open'], high=df['High'], low=df['Low'], close=df['Close'], name='Market'),
    row=1, col=1
)
fig.add_trace(
    go.Bar(x=df['Date'],y=df['Volume'], name='Volume'),
    row=2, col=1
)
RSI = ta.RSI(df['Close'],14)

fig.add_trace(
    go.Scatter(x=df['Date'],y=RSI, name='RSI'),
    row=3, col=1
)
fig.add_hline(y=70, line_dash="dash", line_color="red", row=3, col=1)
fig.add_hline(y=30, line_dash="dash", line_color="green", row=3, col=1)
fig.update_layout(
    width=1400,
    height = 1000,
    xaxis=dict(
        rangeslider=dict(
            visible=False
        )
    )
)

fig.show()
```
