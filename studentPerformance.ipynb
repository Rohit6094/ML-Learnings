```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.graph_objects as go
import plotly.express as px
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

```python

data = pd.read_csv('/home/killua/Desktop/files/ml/data/Student_Performance.csv')
df = pd.DataFrame(data)
df = df.drop_duplicates()
df.info()
scalerST = StandardScaler()
scalerMM = MinMaxScaler()

# df.isnull().sum()
# df.describe()
# df.head(10)
# df.duplicated()
```

```python
# #Gender Mapping/Encoding
# gender_map = {
#     'male': 0,
#     'female': 1,
#     'other': 2
# }
# df['gender'] = df['gender'].map(gender_map)
df = pd.get_dummies(df, columns=['gender','study_method'])

#Encoding Target FInal Grade (Ordinal)
grade_enc = {
    'f' : 0,
    'e' : 1,
    'd' : 2,
    'c' : 3,
    'b' : 4,
    'a' : 5,
}
df['final_grade'] = df['final_grade'].map(grade_enc)


#School Type Mapping
school_type_map = {
    'public' : 0,
    'private' : 1
}
df['school_type'] = df['school_type'].map(school_type_map)


#Parents Education Encoding (Ordinal)
parents_edu_enc = {
    'no formal' : 0,
    'high school' : 1,
    'diploma' : 2,
    'graduate' : 3,
    'post graduate' : 4,
    'phd' : 5
}
df['parent_education'] = df['parent_education'].map(parents_edu_enc)


#Internet access encoding
internet_acc = {
    'yes': 1,
    'no' : 0
}
df['internet_access'] = df['internet_access'].map(internet_acc)


#Travel Time Encoding (Ordinal)
travel_time_enc = {
    '<15 min' : 0,
    '15-30 min' : 1,
    '30-60 min' : 2,
    '>60 min' : 3
}
df['travel_time'] = df['travel_time'].map(travel_time_enc)

#Extra Activities Enc
extra_act_enc = {
    'yes' : 1,
    'no' : 0
}
df['extra_activities'] = df['extra_activities'].map(extra_act_enc)

# df = pd.get_dummies(df, columns=['study_method'])
df.drop(columns=['student_id','math_score','science_score','english_score','overall_score'], inplace=True)

#Scaling stduy_hours and attendance_percentage

# df['study_hours'] = scaler.fit_transform(df[['study_hours']])
# df['attendance_percentage'] = scaler.fit_transform(df[['attendance_percentage']])
df_scaled = scalerMM.fit_transform(df[['study_hours','attendance_percentage']])
df[['study_hours','attendance_percentage']] = df_scaled


df.head(10)
```
